# Transformer + cross-attention recipe reranker

Implements [recipe_reranker_spec.md](recipe_reranker_spec.md). Given a user's rating history and one candidate
recipe from the upstream retrieval stage, the model predicts **P(the user rates it positively)**.

The user is not compressed into one vector. The candidate attends over a static profile token and over the
self-attention state of every rating in the user's recent history. Each part of the model is its own class, as in the
spec:

```text
ItemEncoder      product / adj / verb tokens ─► embedding ─► self-attention ─► mean pool ─► text_emb ─┐
                 [price, nutrients] (log1p, z-score) ─► Linear ─────────────────────────► num_emb ──┴─► Linear(2d, d) ─► item_emb

HistoryEncoder   item_emb + Linear(centered rating in [0, 1]), history before t (date order, last t_max)
                 ─► self-attention (RoPE, GQA, time window) ─► h_1 … h_T ───────────────────────────────────────┐
UserEncoder      avg price + avg nutrients before t ─► standardize ──────┐                                      │
                 avg nutrients before t, rating-weighted ─► standardize ─┴─► weighted avg ──► MLP ─► U_profile ─┴─► keys / values (1+T, d)

TransformerCrossAttentionReranker
                 candidate item_emb ─► query ─► MultiheadAttention ─► attn_out
                 MLP(concat(attn_out, item_emb)) ─► logit ─► sigmoid ─► P(positive)
```

**Where the spec leaves a choice**

| Spec point | Choice made here |
|---|---|
| 1.6 `negative_sampling_power` | 1: sampled negatives are drawn ∝ popularity (ratings in the whole file), so preferring popular recipes does not separate the classes. |
| 2.1 pooling and fusion | Mean pooling with the padding mask; fusion option A; the field embedding is on. |
| 2.2 profile module | Two-layer MLP. Its input is the profile with each nutrient replaced by (x · history average + y · preferred value) / (x + y). The preferred value is that nutrient's mean over the user's ratings before t, each weighted by 1 / 2^(5 − rating), rounded to the nearest integer; the loading cell computes it, as `sampling_preference.ipynb` did. x and y are learned with the rest of the model, equal at the start. Price passes through. |
| 2.3 history encoder | Pre-norm self-attention (`history_layers` layers) over the last `t_max` ratings, not a GRU. RoPE rotates queries and keys by position; `n_kv_heads` (2) key/value heads are each shared by a group of query heads (GQA); a rating only attends to ratings at most `attention_window_days` (183) days apart. |
| 2.3 history rating | `Linear(1, d)` of the centered rating mapped to [0, 1] ((centered + 5) / 10), added to the step's `item_emb`. |
| 3.2 optimizer | AdamW; early stopping on validation log loss. |

````python
from dataclasses import asdict, dataclass
from functools import lru_cache
import json
import math
from pathlib import Path
import random
import re
import time

import matplotlib.pyplot as plt
import nltk
from nltk.stem import WordNetLemmatizer
import numpy as np
import pandas as pd
import torch
from IPython.display import display
from torch import nn
from torch.nn import functional as F
from tqdm.auto import tqdm


@dataclass(frozen=True)
class RerankerConfig:
    seed: int = 42
    max_users: int | None = None          # A random subset of users for a quick run; None keeps everyone.
    positive_min_rating: int = 3          # 3-5 positive, 0-2 negative (spec 1.2).
    min_user_ratings: int = 6             # Spec 1.3.
    negative_sampling_power: float = 1.0  # Sampled negatives ∝ popularity ** power; 0 is uniform.
    min_token_count: int = 3              # Tokens in fewer training recipes map to <unk>.
    max_tokens: int = 40                  # Tokens kept per recipe (p99 is 31).
    d: int = 64
    n_heads: int = 4
    text_layers: int = 1
    t_max: int = 20                       # Most recent history steps fed to the history Transformer.
    history_layers: int = 1               # Self-attention layers over the history.
    n_kv_heads: int = 2                   # Key/value heads of the history self-attention, shared by query heads (GQA).
    attention_window_days: int = 183      # History ratings at most this many days apart attend to each other.
    dropout: float = 0.1                  # History self-attention, cross-attention and head.
    text_dropout: float = 0.0             # Inside the token self-attention: on CPU it costs more than the rest of a step.
    lr: float = 1e-3
    weight_decay: float = 1e-5
    batch_size: int = 512
    chunk_size: int = 16                  # Consecutive rows of one user kept together in a batch.
    max_epochs: int = 10
    patience: int = 2                     # Epochs without a validation log-loss improvement before stopping.
    ranking_negatives: int = 99
    ndcg_k: int = 10


CONFIG = RerankerConfig()

NUTRIENT_COLUMNS = ["calories (g)", "total fat (g)", "sugar (g)", "sodium (g)", "protein (g)",
                    "saturated fat (g)", "carbohydrates (g)"]
PREFERENCE_COLUMNS = [f"preferred {column}" for column in NUTRIENT_COLUMNS]  # Computed when the data is loaded.
PRICE_COLUMN = "total_price"  # The sum of the priced ingredients' prices.
TEXT_FIELDS = {"product": ("n", 1), "adj": ("a", 2), "verb": ("v", 3)}  # WordNet part of speech, field id.
POSITIVE, RATED_NEGATIVE, SAMPLED_NEGATIVE = "positive (3-5)", "rated negative (0-2)", "sampled negative"

random.seed(CONFIG.seed)
np.random.seed(CONFIG.seed)
torch.manual_seed(CONFIG.seed)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")
print(asdict(CONFIG))
````

````python
import pandas as pd

ARTIFACT_DIR = Path(
    "./model_artifacts/transformer_xattn_reranker"
)
INPUT_DIR = "../../input_stage2"

# foodcom_map_recipe_review_price.ipynb writes this file: every Food.com rating joined with its recipe's ingredients,
# product / adj / verb terms, nutrients and price. Only the columns used here are read: the file also holds each
# recipe's description, steps and raw ingredient lines (2.6 GB in all).
USE_COLUMNS = ["user_id", "recipe_id", "rating", "date", "name", *NUTRIENT_COLUMNS, PRICE_COLUMN, *TEXT_FIELDS]
recipe_user_interaction_df = pd.read_csv(INPUT_DIR + "/foodcom_recipe_ratings_ingredients.csv", usecols=USE_COLUMNS)

# Preferred nutrients, as sampling_preference.ipynb added them to the old input: per user, each nutrient's mean over
# all their reviews, a review weighted by 1 / 2^(5 - rating) (5 -> 1, 4 -> 1/2, ..., 0 -> 1/32), rounded to the nearest
# integer, a half going up. Every review of a user carries the same values.
review_weight = 0.5 ** (5 - recipe_user_interaction_df["rating"].to_numpy(np.float64))
review_user = recipe_user_interaction_df["user_id"].to_numpy()
weighted_sum = (pd.DataFrame(recipe_user_interaction_df[NUTRIENT_COLUMNS].to_numpy(np.float64) * review_weight[:, None])
                .groupby(review_user).transform("sum").to_numpy())
weight_sum = pd.Series(review_weight).groupby(review_user).transform("sum").to_numpy()
recipe_user_interaction_df[PREFERENCE_COLUMNS] = np.floor(weighted_sum / weight_sum[:, None] + 0.5)
print(f"{len(recipe_user_interaction_df):,} ratings by {recipe_user_interaction_df['user_id'].nunique():,} users "
      f"of {recipe_user_interaction_df['recipe_id'].nunique():,} recipes")
````

## Phase 1: Preprocessing and profiling

### 1.1 Text normalization

Each word of `product`, `adj` and `verb` is lowercased and stripped of punctuation (a hyphen splits words:
`hard-boiled` → `hard boiled`). It is then lemmatized with WordNet using its column's part of speech (`ground` as
a verb → `grind`) and singularized with WordNet's noun lemmatizer. WordNet only rewrites plurals it knows, so
`hummus`, `couscous` and `molasses` survive; suffix rules would cut them. Each (token, field) pair is kept once
per recipe, in order.

````python
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
_lemmatizer = WordNetLemmatizer()
_NON_ALPHANUMERIC = re.compile(r"[^a-z0-9]+")


@lru_cache(maxsize=None)
def normalize_word(word, pos):
    """Lemmatize with the column's part of speech, then singularize (WordNet noun lemma)."""
    return _lemmatizer.lemmatize(_lemmatizer.lemmatize(word, pos), "n")


def recipe_tokens(product, adj, verb):
    """(token, field id) pairs of one recipe, in order, each kept once."""
    tokens = {}
    for (pos, field), raw in zip(TEXT_FIELDS.values(), (product, adj, verb)):
        for ingredient in json.loads(raw):
            for phrase in ingredient:
                for word in _NON_ALPHANUMERIC.sub(" ", phrase.lower()).split():
                    tokens.setdefault((normalize_word(word, pos), field), None)
    return list(tokens)


# Recipe columns are constant per recipe_id, so one row per recipe is the catalog.
recipes = (recipe_user_interaction_df
           .drop_duplicates("recipe_id")[["recipe_id", "name", *NUTRIENT_COLUMNS, PRICE_COLUMN, *TEXT_FIELDS]]
           .sort_values("recipe_id").reset_index(drop=True))
recipes["tokens"] = [recipe_tokens(*row) for row in recipes[list(TEXT_FIELDS)].itertuples(index=False)]
recipe_index = pd.Series(recipes.index, index=recipes["recipe_id"])
token_counts = recipes["tokens"].str.len()
print(f"{len(recipes):,} recipes; tokens per recipe: median {token_counts.median():.0f}, "
      f"p99 {token_counts.quantile(0.99):.0f}, max {token_counts.max()}")
display(recipes[["recipe_id", "name", *TEXT_FIELDS, "tokens"]].head(3))
````

### 1.2–1.4 Labels, activity filter and rating centering

Ratings 3–5 are positive, 0–2 negative. Users with fewer than 6 observed ratings are dropped, then `mu_user` and
`rating_centered` are computed on the rows that remain. In Food.com a rating of 0 usually means a review without
stars; the spec counts it as negative, and so does this notebook.

````python
interactions = recipe_user_interaction_df[["user_id", "recipe_id", "rating", "date", *PREFERENCE_COLUMNS]].copy()
interactions["date"] = pd.to_datetime(interactions["date"])
interactions["label"] = (interactions["rating"] >= CONFIG.positive_min_rating).astype(np.int8)

ratings_per_user = interactions.groupby("user_id").size()
active_users = ratings_per_user.index[ratings_per_user >= CONFIG.min_user_ratings]
if CONFIG.max_users is not None and len(active_users) > CONFIG.max_users:
    active_users = np.random.default_rng(CONFIG.seed).choice(active_users, CONFIG.max_users, replace=False)
observed = (interactions[interactions["user_id"].isin(active_users)]
            .sort_values(["user_id", "date", "recipe_id"]).reset_index(drop=True))
observed["recipe_idx"] = recipe_index.loc[observed["recipe_id"]].to_numpy()
observed["source"] = np.where(observed["label"] == 1, POSITIVE, RATED_NEGATIVE)
# The filter drops whole users, so this is each kept user's mean either way. Only profiled below: the model
# never sees it.
observed["mu_user"] = observed.groupby("user_id")["rating"].transform("mean")
observed["rating_centered"] = observed["rating"] - observed["mu_user"]
print(f"Activity filter kept {observed['user_id'].nunique():,} of {interactions['user_id'].nunique():,} users "
      f"and {len(observed):,} of {len(interactions):,} ratings")
display(observed.groupby("rating").agg(rows=("label", "size"), label=("label", "first")).T)
````

### 1.5 Split and 1.6 negative sampling

The split comes first so that each sampled negative can inherit its positive's split. Each user's observed ratings
are ordered by date (recipe id breaks ties): the last is test, the second-to-last validation, the rest train. Then
every positive gets one recipe that user never rated, drawn ∝ popularity (ratings in the whole file), with label 0 and the
positive's date. Draws for one user are distinct. A sampled negative has no rating: 0 already means an observed
negative, so its `rating`, `mu_user` and `rating_centered` stay empty.

````python
from_end = observed.groupby("user_id").cumcount(ascending=False)
observed["split"] = np.select([from_end == 0, from_end == 1], ["test", "validation"], "train")

# Ratings per recipe in the whole file: a sampling distribution only, never a model input (spec 1.6).
popularity = np.bincount(recipe_index.loc[interactions["recipe_id"]].to_numpy(), minlength=len(recipes))
sampling_weights = popularity.astype(np.float64) ** CONFIG.negative_sampling_power
sampling_weights /= sampling_weights.sum()
rated_pairs = pd.MultiIndex.from_arrays([observed["user_id"], observed["recipe_idx"]])


def sample_unrated(user_ids, rng):
    """One recipe index per entry of user_ids, drawn ∝ sampling_weights among recipes that user never rated.
    Draws for the same user are distinct."""
    sampled = rng.choice(len(recipes), size=len(user_ids), p=sampling_weights)
    while True:
        pairs = pd.MultiIndex.from_arrays([user_ids, sampled])
        clash = pairs.isin(rated_pairs) | pairs.duplicated()
        if not clash.any():
            return sampled
        sampled[clash] = rng.choice(len(recipes), size=int(clash.sum()), p=sampling_weights)


sampled = observed.loc[observed["label"] == 1, ["user_id", "date", "split"]].copy()
sampled["recipe_idx"] = sample_unrated(sampled["user_id"].to_numpy(), np.random.default_rng(CONFIG.seed))
sampled["recipe_id"] = recipes["recipe_id"].to_numpy()[sampled["recipe_idx"]]
sampled["label"] = np.int8(0)  # No rating, mu_user or rating_centered: they stay NaN (spec 1.6).
sampled["source"] = SAMPLED_NEGATIVE
# Every (user, recipe) pair is unique here, so this order is deterministic.
examples = (pd.concat([observed, sampled], ignore_index=True)
            .sort_values(["user_id", "date", "recipe_idx"]).reset_index(drop=True))
print(f"{len(examples):,} (user, recipe) rows: {len(observed):,} observed and {len(sampled):,} sampled negatives; "
      f"positive share {examples['label'].mean():.1%}")
````

## Phase 2: Examples, normalization and history

Each row of `examples` is one (user, candidate recipe) pair at day t. Its history is the user's observed ratings
on days **strictly before** t (constraint 1): the last `t_max` of them feed the history Transformer, and all of them feed the
profile and, weighted by rating, the preferred nutrient values computed when the data is loaded. Each history step
also carries its rating, centered on the user's mean rating over those earlier days
and mapped to [0, 1] as (centered + 5) / 10, and its day: the history Transformer only lets ratings close in time
attend to each other. Rows with no earlier day (each user's first day) are dropped. The token vocabulary, the item
[price, nutrients] z-score and the profile z-score (which also scales the preferred nutrients) are all fit on
training rows only (constraint 3).

````python
users = pd.Index(observed["user_id"].unique())
DAY_SPAN = 1 << 20  # Day numbers stay below this, so user * DAY_SPAN + day sorts by user, then day.


def user_day_keys(frame):
    days = frame["date"].to_numpy().astype("datetime64[D]").astype(np.int64)
    return users.get_indexer(frame["user_id"]).astype(np.int64) * DAY_SPAN + days


observed_keys = user_day_keys(observed)  # Ascending: observed is sorted by user, then date.
example_keys = user_day_keys(examples)
user_start = np.searchsorted(observed_keys, example_keys - example_keys % DAY_SPAN)
prior_end = np.searchsorted(observed_keys, example_keys)  # First observed rating on or after the row's day.
n_prior = prior_end - user_start
has_history = n_prior > 0
print(f"Dropped {(~has_history).sum():,} rows with no earlier day of history")
examples = examples[has_history].reset_index(drop=True)
user_start, prior_end, n_prior = user_start[has_history], prior_end[has_history], n_prior[has_history]
split = examples["split"].to_numpy()

# Vocabulary from recipes in training rows (targets, sampled negatives and, through them, histories).
train_recipes = np.unique(examples.loc[split == "train", "recipe_idx"])
token_recipe_counts = pd.Series([token for tokens in recipes["tokens"].iloc[train_recipes]
                                 for token in dict.fromkeys(token for token, _ in tokens)]).value_counts()
vocabulary = ["<pad>", "<unk>", *sorted(token_recipe_counts.index[token_recipe_counts >= CONFIG.min_token_count])]
token_lookup = {token: index for index, token in enumerate(vocabulary)}
token_ids = np.zeros((len(recipes), CONFIG.max_tokens), dtype=np.int64)
field_ids = np.zeros_like(token_ids)
for row, tokens in enumerate(recipes["tokens"]):
    tokens = tokens[:CONFIG.max_tokens] or [("<unk>", TEXT_FIELDS["product"][1])]  # An empty recipe gets one <unk>.
    token_ids[row, :len(tokens)] = [token_lookup.get(token, 1) for token, _ in tokens]
    field_ids[row, :len(tokens)] = [field for _, field in tokens]

# Item numbers: log1p, then z-score with statistics of the training recipes. Column 0 is price.
log_numeric = np.log1p(recipes[[PRICE_COLUMN, *NUTRIENT_COLUMNS]].to_numpy(np.float64))
item_mean, item_std = log_numeric[train_recipes].mean(0), log_numeric[train_recipes].std(0)
recipe_numeric = ((log_numeric - item_mean) / item_std).astype(np.float32)

# History: the last t_max observed recipes before the row's day, left-aligned, padded with -1.
observed_recipes = observed["recipe_idx"].to_numpy()
history_len = np.minimum(n_prior, CONFIG.t_max)
steps = np.arange(CONFIG.t_max)
positions = np.clip(prior_end[:, None] - history_len[:, None] + steps, 0, len(observed) - 1)
history_idx = np.where(steps < history_len[:, None], observed_recipes[positions], -1)

# History ratings (spec 2.3): centered on the user's mean rating before the row's day. Not mu_user, which also
# averages the target and later ratings. Ratings are 0-5, so a centered rating lies in [-5, 5], and
# (centered + offset) / scale maps it to [0, 1]. Fixed bounds: nothing to fit, and a per-user std would be 0 for
# users who give every recipe the same rating. Padded steps stay 0.
observed_ratings = observed["rating"].to_numpy(np.float64)
running_rating = np.concatenate([[0.0], np.cumsum(observed_ratings)])
mu_before = (running_rating[prior_end] - running_rating[user_start]) / n_prior
real_step = steps < history_len[:, None]
history_centered = np.where(real_step, observed_ratings[positions] - mu_before[:, None], 0.0)
history_rating_offset, history_rating_scale = 5.0, 10.0
history_rating = np.where(real_step, (history_centered + history_rating_offset) / history_rating_scale,
                          0.0).astype(np.float32)

# History days: two history ratings attend to each other only when they are at most attention_window_days apart.
# Padded steps stay 0.
observed_days = observed["date"].to_numpy().astype("datetime64[D]").astype(np.int64)
history_day = np.where(real_step, observed_days[positions], 0).astype(np.int32)

# Profile: mean log1p [price, nutrients] over every earlier rating, then z-scored on training rows.
running = np.vstack([np.zeros((1, log_numeric.shape[1])), np.cumsum(log_numeric[observed_recipes], axis=0)])
profile_raw = (running[prior_end] - running[user_start]) / n_prior[:, None]
# Price (column 0) is z-scored; the nutrients are min-max scaled. Both fit on training rows.
train_profile = profile_raw[split == "train"]
profile_shift = np.concatenate([train_profile[:, :1].mean(0), train_profile[:, 1:].min(0)])
profile_scale = np.concatenate([train_profile[:, :1].std(0), train_profile[:, 1:].max(0) - train_profile[:, 1:].min(0)])
profile = ((profile_raw - profile_shift) / profile_scale).astype(np.float32)

# Preferred nutrients, from the loading cell: every observed row carries its user's nutrient means over all
# their reviews, each rating weighted by 1 / 2^(5 - rating) and each mean rounded to the nearest integer. An example
# shares its day with an observed row (sampled negatives take their positive's date), and observed row prior_end is
# the first of that day. Put on the profile's log1p scale and scaled with its statistics, so UserEncoder can average
# the two column by column.
preferences_raw = np.log1p(observed[PREFERENCE_COLUMNS].to_numpy(np.float64)[prior_end])
if np.isnan(preferences_raw).any():
    raise ValueError("Rows with earlier ratings but no preferred nutrients: check the preferences in the loading cell")
preferences_nutrient = ((preferences_raw - profile_shift[1:]) / profile_scale[1:]).astype(np.float32)


def split_arrays(name):
    rows = split == name
    # copy=True: pandas copy-on-write hands out read-only arrays, which torch.from_numpy warns about.
    return {"candidate": examples.loc[rows, "recipe_idx"].to_numpy(np.int64, copy=True),
            "history": history_idx[rows], "history_rating": history_rating[rows], "history_day": history_day[rows],
            "history_len": history_len[rows].astype(np.int64),
            "profile": profile[rows], "preferences_nutrient": preferences_nutrient[rows],
            "label": examples.loc[rows, "label"].to_numpy(np.float32, copy=True)}


data = {name: split_arrays(name) for name in ("train", "validation", "test")}
data["train"]["user"] = users.get_indexer(examples.loc[split == "train", "user_id"])
data["train"]["position"] = examples[split == "train"].groupby("user_id").cumcount().to_numpy()
print(f"Vocabulary: {len(vocabulary):,} tokens (appearing in at least {CONFIG.min_token_count} training recipes); "
      f"{(token_ids == 1).sum() / (token_ids > 0).sum():.1%} of catalog tokens map to <unk>")
print(f"History ratings: centered on the mean before t, then (centered + {history_rating_offset:g}) / "
      f"{history_rating_scale:g} in [0, 1]")
# How much of its history a rating attends to (itself included), on the validation rows.
validation_day, validation_real = history_day[split == "validation"], real_step[split == "validation"]
attended = ((np.abs(validation_day[:, :, None] - validation_day[:, None, :]) <= CONFIG.attention_window_days)
            & validation_real[:, None, :]).sum(2)
attended_share = (attended / history_len[split == "validation"][:, None])[validation_real]
print(f"History attention: a rating attends to {attended[validation_real].mean():.1f} history ratings on average, "
      f"{attended_share.mean():.0%} of its row's history (those within {CONFIG.attention_window_days} days)")
split_summary = (examples.groupby(["split", "source"]).size().unstack(fill_value=0)
                 .reindex(index=["train", "validation", "test"], columns=list(SOURCE_COLORS)))
split_summary["positive_share"] = examples.groupby("split")["label"].mean()
split_summary["mean_history_len"] = pd.Series(history_len).groupby(split).mean()
display(split_summary.round(3))
````

### 2.1–2.3 and 3.1: the model

One class per spec section:

- **`ItemEncoder`** (2.1): token + field embeddings → `TransformerEncoderLayer` self-attention (no positional
  encoding: ingredients have no order) → mean pool over real tokens → `text_emb`. z-scored log [price, nutrients]
  → `Linear(8, d)` → `num_emb`. `Linear(2d, d)` over the concat → `item_emb`.
- **`UserEncoder`** (2.2): standardized [mean log price, mean log nutrients], each nutrient replaced by
  (x · it + y · the user's preferred value of it) / (x + y) → two-layer MLP → `U_profile`. The preferred value
  is the nutrient's mean over the earlier ratings, each weighted by 1 / 2^(5 − rating), rounded to the nearest
  integer (computed when the data is loaded). x and y are two learned weights, kept positive and
  equal at the start; the BCE loss trains them with the rest of the model.
- **`HistoryEncoder`** (2.3): the last `t_max` item embeddings in date order, each plus `Linear(1, d)` of its
  normalized centered rating → `history_layers` pre-norm self-attention layers → one state per rating `H`, plus the
  padding mask. The attention differs from `nn.MultiheadAttention` in three ways:
  - **RoPE.** Queries and keys are rotated by the rating's position in the history, so a query-key score depends on
    how many ratings apart the two are, not on where the history starts. No position is added to the tokens.
  - **GQA.** The `n_heads` (4) query heads share `n_kv_heads` (2) key and value heads: each key/value head serves a
    group of two query heads.
  - **Sliding window in time.** A rating attends only to ratings at most `attention_window_days` (183, about six
    months) apart, before or after it, so each state sums up one period of the user's taste. All of the history comes
    before t, so attending to later history ratings leaks nothing.
- **`TransformerCrossAttentionReranker`** (3.1) holds the catalog and the three encoders. Query = candidate `item_emb`;
  keys and values = `[U_profile; h_1 … h_T]` plus a learned type embedding. `key_padding_mask` hides padded steps
  and never the profile token. Head: MLP over `concat(attn_out, item_emb)` → logit; `sigmoid(logit)` = P(positive).
- **Computed in the forward pass.** `forward` gathers every distinct recipe in the batch (candidates and history),
  encodes each once and scatters the embeddings back. Gradients reach the item encoder from the query side and
  through the history Transformer. Training never reads a cache.

````python
class ItemEncoder(nn.Module):
    """Recipe tokens and [price, nutrients] → item_emb (spec 2.1)."""

    def __init__(self, vocab_size, n_numeric, config):
        super().__init__()
        d = config.d
        self.token_embedding = nn.Embedding(vocab_size, d, padding_idx=0)
        self.field_embedding = nn.Embedding(len(TEXT_FIELDS) + 1, d, padding_idx=0)
        layer = nn.TransformerEncoderLayer(d, config.n_heads, dim_feedforward=2 * d, dropout=config.text_dropout,
                                           batch_first=True, norm_first=True)
        self.self_attention = nn.TransformerEncoder(layer, config.text_layers, enable_nested_tensor=False)
        self.numeric_projection = nn.Linear(n_numeric, d)
        self.fusion = nn.Linear(2 * d, d)

    def forward(self, token_ids, field_ids, numeric):
        padding = token_ids == 0
        tokens = self.self_attention(self.token_embedding(token_ids) + self.field_embedding(field_ids),
                                     src_key_padding_mask=padding)
        keep = (~padding).unsqueeze(-1).to(tokens.dtype)
        text_emb = (tokens * keep).sum(1) / keep.sum(1)  # Every recipe has at least one token.
        return self.fusion(torch.cat([text_emb, self.numeric_projection(numeric)], dim=-1))


class UserEncoder(nn.Module):
    """Standardized [mean log price, mean log nutrients] of the ratings before t, each nutrient a learned weighted
    average with the user's preferred value of it → U_profile (spec 2.2)."""

    def __init__(self, n_profile, config):
        super().__init__()
        self.log_weights = nn.Parameter(torch.zeros(2))  # log [x, y]: history average, preferred value. Equal at start.
        self.mlp = nn.Sequential(nn.Linear(n_profile, config.d), nn.ReLU(), nn.Linear(config.d, config.d))

    def forward(self, profile, preferences_nutrient):
        """profile (B, n_profile) is [price, *nutrients]; preferences_nutrient (B, n) holds the preferred values of its
        last n columns, the nutrients, on the same scale. Each of those columns becomes
        (x * profile + y * preferred) / (x + y); price passes through."""
        n = preferences_nutrient.size(1)
        x, y = self.log_weights.exp()
        nutrients = (x * profile[:, -n:] + y * preferences_nutrient) / (x + y)
        return self.mlp(torch.cat([profile[:, :-n], nutrients], dim=1))


def rotate(x, positions, base=10_000.0):
    """RoPE on x (B, heads, T, head_dim): channels i and i + head_dim / 2 turn by positions × base^(-2i / head_dim).
    A rotated query and key then score by the offset between their positions, wherever counting starts."""
    half = x.size(-1) // 2
    angles = positions[:, None] * base ** (-torch.arange(half, device=x.device) / half)  # (T, head_dim / 2)
    cos, sin = angles.cos(), angles.sin()
    first, second = x[..., :half], x[..., half:]
    return torch.cat([first * cos - second * sin, second * cos + first * sin], dim=-1)


class GroupedQueryAttention(nn.Module):
    """Self-attention with RoPE on queries and keys, and n_kv_heads key/value heads each shared by a group of
    n_heads // n_kv_heads query heads (GQA)."""

    def __init__(self, d, n_heads, n_kv_heads, dropout):
        super().__init__()
        if n_heads % n_kv_heads:
            raise ValueError(f"n_heads ({n_heads}) must be a multiple of n_kv_heads ({n_kv_heads})")
        self.n_heads, self.n_kv_heads, self.head_dim, self.dropout = n_heads, n_kv_heads, d // n_heads, dropout
        self.query = nn.Linear(d, n_heads * self.head_dim)
        self.key = nn.Linear(d, n_kv_heads * self.head_dim)
        self.value = nn.Linear(d, n_kv_heads * self.head_dim)
        self.out = nn.Linear(n_heads * self.head_dim, d)

    def forward(self, x, attend):
        """x (B, T, d) → (B, T, d). attend (B, T, T) is True where step i may attend to step j."""
        batch, steps, _ = x.shape
        query = self.query(x).view(batch, steps, self.n_heads, self.head_dim).transpose(1, 2)
        key = self.key(x).view(batch, steps, self.n_kv_heads, self.head_dim).transpose(1, 2)
        value = self.value(x).view(batch, steps, self.n_kv_heads, self.head_dim).transpose(1, 2)
        positions = torch.arange(steps, device=x.device)
        query, key = rotate(query, positions), rotate(key, positions)
        # Query head h reads key/value head h // group.
        group = self.n_heads // self.n_kv_heads
        key, value = key.repeat_interleave(group, dim=1), value.repeat_interleave(group, dim=1)
        out = F.scaled_dot_product_attention(query, key, value, attn_mask=attend.unsqueeze(1),
                                             dropout_p=self.dropout if self.training else 0.0)
        return self.out(out.transpose(1, 2).reshape(batch, steps, -1))


class HistoryLayer(nn.Module):
    """One pre-norm Transformer layer over the history: GroupedQueryAttention, then a feed-forward block."""

    def __init__(self, config):
        super().__init__()
        d = config.d
        self.attention_norm, self.feed_forward_norm = nn.LayerNorm(d), nn.LayerNorm(d)
        self.attention = GroupedQueryAttention(d, config.n_heads, config.n_kv_heads, config.dropout)
        self.feed_forward = nn.Sequential(nn.Linear(d, 2 * d), nn.ReLU(), nn.Dropout(config.dropout),
                                          nn.Linear(2 * d, d))
        self.dropout = nn.Dropout(config.dropout)

    def forward(self, x, attend):
        x = x + self.dropout(self.attention(self.attention_norm(x), attend))
        return x + self.dropout(self.feed_forward(self.feed_forward_norm(x)))


class HistoryEncoder(nn.Module):
    """Item embeddings, normalized centered ratings and days of the history → one self-attention state per rating H
    (spec 2.3)."""

    def __init__(self, config):
        super().__init__()
        self.window_days = config.attention_window_days
        self.rating_projection = nn.Linear(1, config.d)
        self.layers = nn.ModuleList([HistoryLayer(config) for _ in range(config.history_layers)])
        self.norm = nn.LayerNorm(config.d)

    def forward(self, history_emb, history_rating, history_day, history_len):
        """H (B, T, d) and its padding mask (B, T), True = padded.

        history_emb (B, T, d), history_rating (B, T) and history_day (B, T) are left-aligned: steps at or past
        history_len are padding. Each step's token is its item_emb plus a projection of its rating; positions enter
        through RoPE inside the attention. A real step attends to the real steps at most window_days from it, itself
        included. Every step may attend to itself, so a padded step's attention row is never empty."""
        steps = torch.arange(history_emb.size(1), device=history_emb.device)
        padded = steps >= history_len.to(history_emb.device).unsqueeze(1)
        close = (history_day.unsqueeze(2) - history_day.unsqueeze(1)).abs() <= self.window_days
        attend = (close & ~padded.unsqueeze(1)) | (steps.unsqueeze(1) == steps)
        tokens = history_emb + self.rating_projection(history_rating.unsqueeze(-1))
        for layer in self.layers:
            tokens = layer(tokens, attend)
        return self.norm(tokens), padded


class TransformerCrossAttentionReranker(nn.Module):
    """Candidate item_emb cross-attends over [U_profile; H] (spec 3.1)."""

    def __init__(self, token_ids, field_ids, numeric, vocab_size, n_profile, config):
        super().__init__()
        d = config.d
        # The catalog rides along as non-persistent buffers, so forward() takes recipe indices.
        self.register_buffer("token_ids", torch.as_tensor(token_ids), persistent=False)
        self.register_buffer("field_ids", torch.as_tensor(field_ids), persistent=False)
        self.register_buffer("numeric", torch.as_tensor(numeric), persistent=False)
        self.item_encoder = ItemEncoder(vocab_size, numeric.shape[1], config)
        self.user_encoder = UserEncoder(n_profile, config)
        self.history_encoder = HistoryEncoder(config)
        self.key_type = nn.Embedding(2, d)  # 0: profile token, 1: history state.
        self.cross_attention = nn.MultiheadAttention(d, config.n_heads, dropout=config.dropout, batch_first=True)
        self.head = nn.Sequential(nn.Linear(2 * d, d), nn.ReLU(), nn.Dropout(config.dropout), nn.Linear(d, 1))

    def encode_items(self, recipe_idx):
        """item_emb (N, d) for a 1-D tensor of catalog indices."""
        token_ids = self.token_ids[recipe_idx]
        width = int((token_ids != 0).sum(1).max())  # Drop padding columns that no recipe in the batch uses.
        return self.item_encoder(token_ids[:, :width], self.field_ids[recipe_idx, :width], self.numeric[recipe_idx])

    def user_keys(self, history_emb, history_rating, history_day, history_len, profile, preferences_nutrient):
        """Keys and values [U_profile; h_1 … h_T] (B, 1+T, d) and their padding mask (True = padded)."""
        states, padded = self.history_encoder(history_emb, history_rating, history_day, history_len)
        profile_token = self.user_encoder(profile, preferences_nutrient).unsqueeze(1)
        keys = torch.cat([profile_token + self.key_type.weight[0], states + self.key_type.weight[1]], dim=1)
        return keys, torch.cat([padded.new_zeros(len(padded), 1), padded], dim=1)

    def score_candidates(self, candidate_emb, keys, key_padding_mask):
        """Logits (B, C) for candidates (B, C, d). Each candidate attends to its own user's keys only."""
        attended, _ = self.cross_attention(candidate_emb, keys, keys, key_padding_mask=key_padding_mask,
                                           need_weights=False)
        return self.head(torch.cat([attended, candidate_emb], dim=-1)).squeeze(-1)

    def forward(self, candidate_idx, history_idx, history_rating, history_day, history_len, profile,
                preferences_nutrient):
        """One logit per row. Every distinct recipe in the batch is encoded once, in this pass."""
        real = history_idx >= 0
        recipe_idx, slot = torch.unique(torch.cat([candidate_idx, history_idx[real]]), return_inverse=True)
        item_emb = self.encode_items(recipe_idx)
        candidate_emb = item_emb[slot[:len(candidate_idx)]]
        history_emb = item_emb.new_zeros(*history_idx.shape, item_emb.size(1))
        history_emb[real] = item_emb[slot[len(candidate_idx):]]
        keys, key_padding_mask = self.user_keys(history_emb, history_rating, history_day, history_len, profile,
                                                preferences_nutrient)
        return self.score_candidates(candidate_emb.unsqueeze(1), keys, key_padding_mask).squeeze(1)


torch.manual_seed(CONFIG.seed)
model = TransformerCrossAttentionReranker(token_ids, field_ids, recipe_numeric, len(vocabulary), profile.shape[1],
                                          CONFIG).to(device)
print(model)
print(f"{sum(parameter.numel() for parameter in model.parameters()):,} parameters")
````

## 3.2 Training

BCE on the binary label with AdamW. After each epoch the validation log loss decides early stopping
(`patience`), and the best epoch's weights are restored.

A batch is built from chunks of up to `chunk_size` consecutive rows of one user, with the chunks shuffled.
Rows of a chunk share most of their history, so the batch encodes far fewer distinct recipes than
rows × `t_max`. Each row still has its own history and profile; chunking only decides which rows share a batch.

````python
def to_batch(arrays, index):
    """Model inputs for rows `index`, with history trimmed to the batch's longest history."""
    lengths = arrays["history_len"][index]
    return (torch.from_numpy(arrays["candidate"][index]).to(device),
            torch.from_numpy(arrays["history"][index, :lengths.max()]).to(device),
            torch.from_numpy(arrays["history_rating"][index, :lengths.max()]).to(device),
            torch.from_numpy(arrays["history_day"][index, :lengths.max()]).to(device),
            torch.from_numpy(lengths).to(device),
            torch.from_numpy(arrays["profile"][index]).to(device),
            torch.from_numpy(arrays["preferences_nutrient"][index]).to(device))


def epoch_order(user, position, chunk_size, rng):
    """Row order for one epoch: chunks of up to chunk_size consecutive rows of one user, chunks shuffled.
    A random shift per user moves the chunk borders every epoch."""
    shift = rng.integers(0, chunk_size, size=user.max() + 1)[user]
    chunk = user.astype(np.int64) * (1 << 32) + (position + shift) // chunk_size
    _, chunk_id = np.unique(chunk, return_inverse=True)
    return np.argsort(rng.permutation(chunk_id.max() + 1)[chunk_id], kind="stable")


@torch.no_grad()
def predict_proba(model, arrays, batch_size=2048, make_batch=to_batch):
    """P(positive) per row, item embeddings computed by the item encoder. make_batch is as in train_model."""
    model.eval()
    n = len(arrays["label"])
    logits = [model(*make_batch(arrays, np.arange(start, min(start + batch_size, n))))
              for start in range(0, n, batch_size)]
    return torch.sigmoid(torch.cat(logits)).cpu().numpy()


def roc_auc(labels, scores):
    """Probability that a random positive outscores a random negative (ties count half)."""
    labels = np.asarray(labels, dtype=bool)
    ranks = pd.Series(scores).rank().to_numpy()
    n_positive, n_negative = labels.sum(), (~labels).sum()
    return (ranks[labels].sum() - n_positive * (n_positive + 1) / 2) / (n_positive * n_negative)


def classification_metrics(labels, probabilities, threshold=0.5):
    labels = np.asarray(labels, dtype=bool)
    clipped = np.clip(probabilities, 1e-7, 1 - 1e-7)
    predicted = probabilities >= threshold
    true_positive = (predicted & labels).sum()
    precision = true_positive / max(predicted.sum(), 1)
    recall = true_positive / max(labels.sum(), 1)
    return {"log_loss": -np.mean(np.where(labels, np.log(clipped), np.log(1 - clipped))),
            "roc_auc": roc_auc(labels, probabilities), "accuracy": np.mean(predicted == labels),
            "precision": precision, "recall": recall,
            "f1": 2 * precision * recall / (precision + recall) if precision + recall else 0.0}


def train_model(model, train, validation, config, make_batch=to_batch):
    """AdamW on BCE with early stopping on validation log loss; restores the best epoch's weights.
    make_batch(arrays, index) builds the model inputs: to_batch, or window_batch for the window Transformer below."""
    torch.manual_seed(config.seed)
    rng = np.random.default_rng(config.seed)
    optimizer = torch.optim.AdamW(model.parameters(), lr=config.lr, weight_decay=config.weight_decay)
    history, best_loss, best_state, stale = [], math.inf, None, 0
    for epoch in range(1, config.max_epochs + 1):
        started = time.perf_counter()
        model.train()
        order, total_loss = epoch_order(train["user"], train["position"], config.chunk_size, rng), 0.0
        progress = tqdm(range(0, len(order), config.batch_size), desc=f"epoch {epoch}/{config.max_epochs}",
                        unit="batch")
        for start in progress:
            index = order[start:start + config.batch_size]
            logits = model(*make_batch(train, index))
            loss = F.binary_cross_entropy_with_logits(logits, torch.from_numpy(train["label"][index]).to(device))
            optimizer.zero_grad()
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            total_loss += loss.item() * len(index)
            progress.set_postfix(train_log_loss=f"{total_loss / (start + len(index)):.4f}")
        metrics = classification_metrics(validation["label"], predict_proba(model, validation, make_batch=make_batch))
        history.append({"epoch": epoch, "train_log_loss": total_loss / len(order),
                        "validation_log_loss": metrics["log_loss"], "validation_roc_auc": metrics["roc_auc"],
                        "seconds": time.perf_counter() - started})
        print(f"epoch {epoch}: train log loss {history[-1]['train_log_loss']:.4f}, validation log loss "
              f"{metrics['log_loss']:.4f}, validation ROC-AUC {metrics['roc_auc']:.4f} "
              f"({history[-1]['seconds']:.0f}s)")
        if metrics["log_loss"] < best_loss - 1e-4:
            best_loss, stale = metrics["log_loss"], 0
            best_state = {name: value.detach().clone() for name, value in model.state_dict().items()}
        else:
            stale += 1
            if stale >= config.patience:
                break
    model.load_state_dict(best_state)
    return pd.DataFrame(history)
````

````python
training_history = train_model(model, data["train"], data["validation"], CONFIG)
best_epoch = int(training_history.loc[training_history["validation_log_loss"].idxmin(), "epoch"])
print(f"Best epoch {best_epoch}; its weights are restored")
x, y = model.user_encoder.log_weights.exp().tolist()
print(f"UserEncoder nutrient weights: history average {x / (x + y):.1%}, preferred value {y / (x + y):.1%}")
````

## 3.4 Item-embedding cache and serving

After training, `item_emb` is computed once per recipe and stored as `item_cache = {recipe_id: item_emb}`.
Because the cache is built from the final weights, it is never stale. At serving time the user side comes from
the cache: the last `t_max` cached embeddings feed the history Transformer with their normalized centered ratings and days, and the profile comes from the history's prices and
nutrients, the preferred nutrients from all of them weighted by rating.
The check below scores every test row through the cache and requires the forward-pass probabilities.

````python
@torch.no_grad()
def encode_catalog(model, batch_size=4096):
    """item_emb for every recipe, in catalog order."""
    model.eval()
    return torch.cat([model.encode_items(torch.arange(start, min(start + batch_size, len(recipes)), device=device)).cpu()
                      for start in range(0, len(recipes), batch_size)])


catalog_embeddings = encode_catalog(model)
item_cache = dict(zip(recipes["recipe_id"].tolist(), catalog_embeddings))  # {recipe_id: item_emb}


@torch.no_grad()
def score_cached(model, embeddings, arrays, rows, candidates, batch_size=256, make_batch=to_batch):
    """P(positive) (len(rows), C): candidates[i] scored against the history and profile of arrays row rows[i].
    embeddings are the cached item embeddings in catalog order, the values of item_cache; make_batch is as in
    train_model."""
    model.eval()
    scores = []
    for start in range(0, len(rows), batch_size):
        index = rows[start:start + batch_size]
        _, history_idx, history_rating, history_day, history_len, profile, preferences = make_batch(arrays, index)
        history_emb = embeddings[history_idx.clamp(min=0).cpu()].to(device) * (history_idx >= 0).unsqueeze(-1)
        keys, key_padding_mask = model.user_keys(history_emb, history_rating, history_day, history_len, profile,
                                                 preferences)
        candidate_emb = embeddings[torch.from_numpy(candidates[start:start + batch_size])].to(device)
        scores.append(torch.sigmoid(model.score_candidates(candidate_emb, keys, key_padding_mask)).cpu())
    return torch.cat(scores).numpy()


@torch.no_grad()
def rerank_candidates(model, item_cache, user_ratings, candidate_ids, as_of):
    """Rank candidate recipe_ids for one user, most likely positive first, from cached item embeddings.

    user_ratings holds that user's ratings (recipe_id, rating, date). Only days before as_of count: the last t_max
    feed the history Transformer, with their days and their ratings centered on the mean of all of them, and all of
    them feed the profile and, weighted by rating, the preferred nutrients."""
    model.eval()
    earlier = user_ratings[user_ratings["date"] < as_of].sort_values(["date", "recipe_id"])
    if earlier.empty:
        raise ValueError("The reranker needs at least one rating before as_of.")
    recent = earlier["recipe_id"].to_numpy()[-CONFIG.t_max:]
    history_emb = torch.stack([item_cache[recipe_id] for recipe_id in recent]).unsqueeze(0).to(device)
    centered = earlier["rating"].to_numpy(np.float64)[-CONFIG.t_max:] - earlier["rating"].mean()
    history_rating = torch.as_tensor((centered + history_rating_offset) / history_rating_scale, dtype=torch.float32,
                                     device=device).unsqueeze(0)
    days = earlier["date"].to_numpy().astype("datetime64[D]").astype(np.int64)[-CONFIG.t_max:]
    history_day = torch.as_tensor(days, device=device).unsqueeze(0)
    earlier_numeric = log_numeric[recipe_index.loc[earlier["recipe_id"]].to_numpy()]
    profile_raw = earlier_numeric.mean(0)
    # Preferred nutrients as the loading cell computes them: the weighted average of all the user's
    # ratings' nutrients, weights 1 / 2^(5 - rating), rounded to the nearest integer (a half up); then the profile's log1p scale.
    weights = 1 / 2.0 ** (5 - user_ratings["rating"].to_numpy())
    earlier_nutrients = recipes[NUTRIENT_COLUMNS].to_numpy(np.float64)[recipe_index.loc[user_ratings["recipe_id"]]]
    preferences_raw = np.log1p(np.floor(weights @ earlier_nutrients / weights.sum() + 0.5))
    user_profile = torch.as_tensor((profile_raw - profile_shift) / profile_scale, dtype=torch.float32, device=device)
    user_preferences = torch.as_tensor((preferences_raw - profile_shift[1:]) / profile_scale[1:], dtype=torch.float32,
                                       device=device)
    keys, key_padding_mask = model.user_keys(history_emb, history_rating, history_day, torch.tensor([len(recent)]),
                                             user_profile.unsqueeze(0), user_preferences.unsqueeze(0))
    candidate_emb = torch.stack([item_cache[recipe_id] for recipe_id in candidate_ids]).unsqueeze(0).to(device)
    probability = torch.sigmoid(model.score_candidates(candidate_emb, keys, key_padding_mask))[0].cpu().numpy()
    ranked = pd.DataFrame({"recipe_id": candidate_ids, "p_positive": probability})
    ranked = ranked.merge(recipes[["recipe_id", "name"]], on="recipe_id", how="left")
    return ranked.sort_values("p_positive", ascending=False, ignore_index=True)


cached_probability = score_cached(model, catalog_embeddings, test, np.arange(len(test_rows)),
                                  test["candidate"][:, None])[:, 0]
cache_gap = float(np.abs(cached_probability - test_probability).max())
if cache_gap > 1e-4:
    raise RuntimeError(f"Cached scoring differs from the forward pass by up to {cache_gap:.2e}")
print(f"item_cache holds {len(item_cache):,} recipes × {catalog_embeddings.shape[1]} dims; scoring the test rows "
      f"from it matches the forward pass (max gap {cache_gap:.1e})")
````

### Ranking: HR@10 and NDCG@10

Each test user whose last rating is positive gets a list of 100 candidates: that recipe plus 99 recipes the
user never rated, sampled like the training negatives (∝ popularity). All of them are scored from the cache with
the same history. `hit_rate@10` is the share of users whose held-out recipe lands in the top 10, and random order
gives 0.10. Because the candidates are drawn ∝ popularity, popularity alone has little to go on.

````python
def ranking_metrics(scores, k):
    """scores (users, candidates) with the held-out positive in column 0."""
    rank = 1 + (scores[:, 1:] > scores[:, :1]).sum(1)
    return {f"hit_rate@{k}": np.mean(rank <= k),
            f"ndcg@{k}": np.mean(np.where(rank <= k, 1 / np.log2(rank + 1), 0.0)),
            "mrr": np.mean(1 / rank), "mean_rank": rank.mean()}


ranking_rng = np.random.default_rng(CONFIG.seed + 1)
held_out = np.flatnonzero(test_rows["source"].eq(POSITIVE).to_numpy())  # At most one per user.
held_out_users = test_rows["user_id"].to_numpy()[held_out]
ranking_negatives = sample_unrated(np.repeat(held_out_users, CONFIG.ranking_negatives),
                                   ranking_rng).reshape(len(held_out), -1)
candidate_lists = np.column_stack([test["candidate"][held_out], ranking_negatives])
list_scores = score_cached(model, catalog_embeddings, test, held_out, candidate_lists)
jitter = ranking_rng.random(candidate_lists.shape) * 1e-6  # Breaks ties between equal popularity counts.
ranking = pd.DataFrame({
    "transformer_cross_attention": ranking_metrics(list_scores, CONFIG.ndcg_k),
    "popularity": ranking_metrics(np.log1p(train_popularity[candidate_lists]) + jitter, CONFIG.ndcg_k),
    "random": ranking_metrics(ranking_rng.random(candidate_lists.shape), CONFIG.ndcg_k),
}).T
print(f"{len(held_out):,} users, {candidate_lists.shape[1]} candidates each")
display(ranking.round(4))
````

## Comparison: a Transformer over history windows

The same reranker with another history Transformer, run over **history windows** instead of one flat sequence of
the last `t_max` ratings. The item encoder, profile, cross-attention, head, training loop, splits and metrics are the ones
above: both models train on the same batches in the same order, and with the same seed every module but the history
encoder starts from the flat model's initial weights. Run the cells above first: the comparison reads their results.

- **History windows.** A user's ratings before t, in date order, are cut wherever two consecutive ratings are
  `window_gap_days` (183, about six months) or more apart, so each window is one period of the user's taste. The
  last `max_windows` (10) windows are kept, each with its last `t_max` (20) ratings. `window_starts` and
  `history_windows` are copied from the API's
  [model_inference_score_recommend.py](../../api/src/remymy_api/inference/model_inference_score_recommend.py), so
  serving cuts the windows the same way. Ratings are centered on the mean before t and mapped to [0, 1], as for the
  flat model.
- **The inputs differ.** The windows always hold the flat model's last 20 ratings and add older periods; the table
  below counts the rows where they reach further back.
- **So do the encoders.** The flat model's attention has RoPE, GQA and the sliding window in time; the window model's
  is full multi-head self-attention across all of its windows, with learned window positions. A gap between the two
  models is the encoder and the extra history together.

````python
@dataclass(frozen=True)
class WindowTransformerConfig(RerankerConfig):
    """RerankerConfig plus history windows. t_max is now the ratings kept per window. n_kv_heads and
    attention_window_days are not used: the window model's history encoder has full multi-head attention."""
    max_windows: int = 10                 # Most recent windows kept.
    window_gap_days: int = 183            # A gap this long between two ratings (about six months) starts a window.


WINDOW_CONFIG = WindowTransformerConfig()
WINDOW_ARTIFACT_DIR = ARTIFACT_DIR.parent / "window_transformer_reranker"


# window_starts and history_windows are copied from api/src/remymy_api/inference/model_inference_score_recommend.py,
# so serving cuts the history exactly as training does.
def window_starts(user_codes, days, gap_days):
    """Index of the first rating of each rating's window. Rows are sorted by user, then day. A window ends where
    the user changes or where the next rating comes gap_days or more after the previous one."""
    new_window = np.ones(len(days), dtype=bool)
    new_window[1:] = (user_codes[1:] != user_codes[:-1]) | (np.diff(days) >= gap_days)
    return np.maximum.accumulate(np.where(new_window, np.arange(len(days)), 0))


def history_windows(prior_end, user_start, mu_before, first_of_window, recipe_idx, ratings, max_windows, t_max):
    """Each row's history as windows, most recent window first.

    A row's history is ratings[user_start:prior_end]. Returns recipe indices (B, max_windows, t_max), -1 at padding;
    ratings centered on mu_before (B, max_windows, t_max), 0 at padding; window lengths (B, max_windows), 0 for
    missing windows. Each window keeps its last t_max ratings, in date order."""
    steps = np.arange(t_max)
    history_idx = np.full((len(prior_end), max_windows, t_max), -1, dtype=np.int64)
    history_rating = np.zeros((len(prior_end), max_windows, t_max), dtype=np.float32)
    window_len = np.zeros((len(prior_end), max_windows), dtype=np.int64)
    ends = np.asarray(prior_end, dtype=np.int64).copy()
    for k in range(max_windows):
        alive = ends > user_start
        if not alive.any():
            break
        # The window of the last rating before `ends`; it never reaches before user_start (a user change starts one).
        starts = np.where(alive, first_of_window[np.maximum(ends - 1, 0)], ends)
        length = np.minimum(ends - starts, t_max)
        positions = np.clip(ends[:, None] - length[:, None] + steps, 0, len(recipe_idx) - 1)
        real = steps < length[:, None]
        history_idx[:, k] = np.where(real, recipe_idx[positions], -1)
        history_rating[:, k] = np.where(real, ratings[positions] - mu_before[:, None], 0.0)
        window_len[:, k] = length
        ends = starts
    return history_idx, history_rating, window_len


# observed is sorted by user, then date: a rating's window only depends on the ratings up to it.
first_of_window = window_starts(users.get_indexer(observed["user_id"]), observed_days, WINDOW_CONFIG.window_gap_days)


def window_split(name):
    """A split of `data` with each row's range of earlier ratings in place of its flat history. window_batch cuts the
    windows per batch, so a (rows, max_windows, t_max) array of the whole training split never sits in memory."""
    rows = split == name
    kept = {key: values for key, values in data[name].items() if not key.startswith("history")}
    return {**kept, "user_start": user_start[rows], "prior_end": prior_end[rows], "mu_before": mu_before[rows]}


def row_windows(arrays, index):
    """history_windows of rows `index` of a window_data split: recipes, centered ratings and window lengths."""
    return history_windows(arrays["prior_end"][index], arrays["user_start"][index], arrays["mu_before"][index],
                           first_of_window, observed_recipes, observed_ratings, WINDOW_CONFIG.max_windows,
                           WINDOW_CONFIG.t_max)


def window_batch(arrays, index):
    """Model inputs for rows `index` of a window_data split: history (B, K, T) with its ratings mapped to [0, 1] as for
    the flat model, no days (the windows already group the ratings by time) and window lengths (B, K), trimmed to the
    batch's windows and its longest window."""
    history, centered, window_len = row_windows(arrays, index)
    windows, steps = int((window_len > 0).sum(1).max()), int(window_len.max())
    history, centered = history[:, :windows, :steps], centered[:, :windows, :steps]
    rating = np.where(history >= 0, (centered + history_rating_offset) / history_rating_scale, 0.0)
    return (torch.from_numpy(arrays["candidate"][index]).to(device),
            torch.from_numpy(np.ascontiguousarray(history)).to(device),
            torch.from_numpy(rating.astype(np.float32)).to(device),
            None,  # history_day: the window model does not use it.
            torch.from_numpy(np.ascontiguousarray(window_len[:, :windows])).to(device),
            torch.from_numpy(arrays["profile"][index]).to(device),
            torch.from_numpy(arrays["preferences_nutrient"][index]).to(device))


window_data = {name: window_split(name) for name in data}
window_summary = {}
for name, arrays in window_data.items():
    window_len = np.concatenate([row_windows(arrays, slice(start, start + 65536))[2]
                                 for start in range(0, len(arrays["label"]), 65536)])
    n_windows, in_windows, flat_sees = (window_len > 0).sum(1), window_len.sum(1), data[name]["history_len"]
    window_summary[name] = {"mean_windows": n_windows.mean(), "rows_with_2+_windows": np.mean(n_windows >= 2),
                            "mean_ratings_in_windows": in_windows.mean(), "mean_ratings_flat_sees": flat_sees.mean(),
                            "rows_with_ratings_older_than_flat": np.mean(in_windows > flat_sees)}
print(f"Windows: a gap of {WINDOW_CONFIG.window_gap_days}+ days starts one; the last {WINDOW_CONFIG.max_windows} "
      f"are kept, with the last {WINDOW_CONFIG.t_max} ratings of each")
display(pd.DataFrame(window_summary).T.round(3))
````

### The Transformer history encoder

- **Tokens.** Each history rating is one token: `item_emb + Linear(rating)`, as in the flat model, plus two learned
  positional encodings (the flat model uses RoPE instead): the **window position** (0 = the most recent window) and
  the **sub-position** inside the window (0 = the window's most recent rating). Both count back from the most recent
  rating, so they mean the same for a user with 3 ratings as for one with 300.
- **Windows attend to each other.** The tokens of all windows form one sequence for a pre-norm `TransformerEncoder`
  (`history_layers` layers, padding masked): each rating attends to the ratings of its own window and of every other
  window. Each row's tokens are first packed to the left, so attention skips the padding of short windows.
- **Reranker.** `WindowTransformerReranker` subclasses `TransformerCrossAttentionReranker` and only swaps
  `history_encoder`, so `forward`, `user_keys` and `score_candidates` are reused: the candidate cross-attends over
  `[U_profile; one state per history rating]` as in the flat model. `window_batch` passes `None` for the history
  days, which this encoder does not use.

````python
class WindowTransformerHistoryEncoder(nn.Module):
    """History windows → one self-attention state per history rating; takes HistoryEncoder's place in the reranker."""

    def __init__(self, config):
        super().__init__()
        d = config.d
        self.rating_projection = nn.Linear(1, d)
        self.window_position = nn.Embedding(config.max_windows, d)  # 0: the most recent window.
        self.sub_position = nn.Embedding(config.t_max, d)  # Inside a window, 0: its most recent rating.
        layer = nn.TransformerEncoderLayer(d, config.n_heads, dim_feedforward=2 * d, dropout=config.dropout,
                                           batch_first=True, norm_first=True)
        self.self_attention = nn.TransformerEncoder(layer, config.history_layers, norm=nn.LayerNorm(d),
                                                    enable_nested_tensor=False)

    def forward(self, history_emb, history_rating, history_day, window_len):
        """States (B, L, d) and their padding mask (B, L), True = padded: row b's ratings fill its first
        window_len[b].sum() slots, and L is the most that any row of the batch has.

        history_emb (B, K, T, d) and history_rating (B, K, T) hold window k's ratings in date order, left-aligned,
        window 0 the most recent; window_len (B, K) counts them, 0 for a window the user does not have. The tokens of
        all K windows form one sequence, so each rating attends within its own window and across the others.
        history_day is not used (None): the windows already group the ratings by time."""
        window_len = window_len.to(history_emb.device)
        windows, steps = history_emb.shape[1:3]
        steps_back = window_len.unsqueeze(-1) - 1 - torch.arange(steps, device=history_emb.device)  # (B, K, T)
        real = steps_back >= 0
        tokens = (history_emb + self.rating_projection(history_rating.unsqueeze(-1))
                  + self.window_position.weight[:windows, None] + self.sub_position(steps_back.clamp(min=0)))
        # Each row's real tokens move to its first slots: self-attention then skips the padding of short windows.
        # Both sides of the assignment list rows in order, each row's tokens in (window, step) order.
        count = window_len.sum(1)
        padded = torch.arange(int(count.max()), device=history_emb.device) >= count.unsqueeze(1)
        packed = tokens.new_zeros(*padded.shape, tokens.size(-1))
        packed[~padded] = tokens[real]
        return self.self_attention(packed, src_key_padding_mask=padded), padded


class WindowTransformerReranker(TransformerCrossAttentionReranker):
    """TransformerCrossAttentionReranker over history windows, with WindowTransformerHistoryEncoder in place of
    HistoryEncoder.

    forward, user_keys and score_candidates are inherited: they take history (B, K, T), no days and window lengths
    (B, K) in place of history (B, T), its days and its lengths, and the candidate attends over
    [U_profile; one state per rating]."""

    def __init__(self, token_ids, field_ids, numeric, vocab_size, n_profile, config):
        super().__init__(token_ids, field_ids, numeric, vocab_size, n_profile, config)
        self.history_encoder = WindowTransformerHistoryEncoder(config)


def parameter_count(module):
    return sum(parameter.numel() for parameter in module.parameters())


# The same seed as the flat model: every module but the history encoder starts from the flat model's initial weights.
torch.manual_seed(WINDOW_CONFIG.seed)
window_model = WindowTransformerReranker(token_ids, field_ids, recipe_numeric, len(vocabulary), profile.shape[1],
                                         WINDOW_CONFIG).to(device)
print(window_model.history_encoder)
print(f"{parameter_count(window_model):,} parameters, {parameter_count(window_model.history_encoder):,} of them in "
      f"the history encoder (flat model: {parameter_count(model):,} and {parameter_count(model.history_encoder):,})")
````

better lower, every other column higher; the last row is window − flat.
